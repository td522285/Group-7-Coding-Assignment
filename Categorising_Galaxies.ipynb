# Categorising Galaxies
Let's have a look at what galaxies are 'elliptical' and how many we have.

We do by Visual Morphology and by LambdaR_RE parameter.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## By Visual Morphology
In this section, we'll choose elliptical galaxies based on Visual Morphology rating of 0 or 0.5, and see how many we have. 

These will only include galaxies which are part of 
 - VisualMorphology
 - CatClusters
 - StelKin

```python
# Reusing 'Conservative_Elliptical_Raw' code for this first part.
# I have renamed 'elliptical' to 'el_zero'.
samimorph = pd.read_csv('SAMI_Visual_Morphology/samiDR3VisualMorphology.csv')
samicat = pd.read_csv('SAMI_Input_Cat_Cluster/samiDR3InputCatClusters.csv')
samikin = pd.read_csv('SAMI_Kinematics/samiDR3StelKin.csv')

print('Sample Size of SAMI Visual Morphology Dataset:', len(samimorph))
print('Sample Size of SAMI Input Catalogue Dataset:', len(samicat))
print('Sample Size of SAMI Stellar Kinematics Dataset:', len(samikin))

morphcat = pd.merge(samimorph, samicat, on='catid', how='inner')
df = pd.merge(morphcat, samikin, on='catid', how='inner')
print('Final Data Set', len(df))
#print(df.head(20))

el_zero = df[df['type'] == 0.0] # choosing only elliptical galaxies
# key defines elliptical galaxies as having the type = 0
    # 0.5 is for galaxies between elliptical and lenticular stages

print('Number of Elliptical Galaxies:', len(el_zero))
#print('Elliptical Galaxies:', el_zero.head(20))
```

```python
# Now, with a Visual Morphology of 0.5.
el_half = df[df['type'] == 0.5]
print(f"Number of el_half galaxies, {len(el_half)}");
```

## By LambdaR_Re
Elliptical galaxies may actually have high mean velocity. These galaxies are often split into 'slow-rotators' or 'fast rotators'. However, we must only look at slow-rotators (low mean velocity) to measure the Faber-Jackson relation. 

I could not find a good velocity/velocity_dispersion benchmark, but I could for the lambda parameter. This measures the amount of angular momentum that supports ordered motion rather than random motion. Emsellem et al. (2011) and van de Sande et al. (2017) support that lambda <= 0.1 is a good benchmark for slow-rotators.

This will only include galaxies that are part of
 - CatClusters
 - StelKin

StelKin has four lambdaR parameters: LAMBDAR_RE, LAMBDAR_RMAX, LAMBDAR_REMGE, LAMBDAR_MAX_MGE. We'll use LAMBDAR_RE as the velocity dispersion at RE is more representative than at RMAX, and a quick scan through each galaxy shows similar values of LAMBDAR_RE vs LAMBDAR_REMGE.

```python
samicat = pd.read_csv('SAMI_Input_Cat_Cluster/samiDR3InputCatClusters.csv')
stelkin = pd.read_csv('SAMI_Kinematics/samiDR3StelKin.csv')
df = pd.merge(samicat, samikin, on='catid');
print(f"The number of galaxies similar to samicat and stelkin is {len(df)}");

slowr = df[df['lambdar_re'] <= 0.1];
print(f"The number of slow-rotators, given by lambdaR_Re <= 0.1, is {len(slowr)}");
```

# Bulge-to-Total
Finally, a high bulge-to-total ratio implies an elliptical galaxy over a disc galaxy.

The benchmark here will be B/T > 0.8, which is provided in the PHYS3116 slides.

This will only include galaxies that are part of
 - CatClusters
 - StelKin

```python
samicat = pd.read_csv('SAMI_Input_Cat_Cluster/samiDR3InputCatClusters.csv')
stelkin = pd.read_csv('SAMI_Kinematics/samiDR3StelKin.csv')
df = pd.merge(samicat, samikin, on='catid');
print(f"The number of galaxies similar to samicat and stelkin is {len(df)}");

slowr = df[df['lambdar_re'] <= 0.1];
print(f"The number of slow-rotators, given by lambdaR_Re <= 0.1, is {len(slowr)}");
```
